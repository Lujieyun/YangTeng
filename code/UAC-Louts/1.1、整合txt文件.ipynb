```python
import os
import pandas as pd
import json

list_file = list(os.walk('C:/Users/YangTeng/Downloads/爬虫/UAC/menu'))[0][2]
if '.DS_Store' in list_file:
   list_file.remove('.DS_Store')

len(list_file)

df = pd.DataFrame(columns=['Part_Number',
                           'Url',
                           'detail'])

df

for file in list_file:
    print(file)
    with open('C:/Users/YangTeng/Downloads/爬虫/UAC/menu/' + file,encoding='utf-8') as f:
        list_part = f.read()[1:-1].split('},{')
        for i in range(len(list_part)):
            if i == 0:
                part = json.loads(list_part[i] + '}')
            elif i == len(list_part) - 1:
                part = json.loads('{' + list_part[i])
            else:
                part = json.loads('{' + list_part[i] + '}')
                
            df_temp = pd.DataFrame([{'Part_Number': part['Part_Number'],
                                     'Url': part['Url'],
                                     'detail': part['detail']
                                     }])
                                     
            df = pd.concat([df, df_temp], ignore_index=True)
            
df.to_excel('C:/Users/YangTeng/Downloads/爬虫/UAC/menu.xlsx', index=False)

df
```
