```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from lxml import etree

ua = pd.read_excel(r'C:\Users\YangTeng\Desktop\UAC-Compressor Assembly\menu.xlsx', header=0)
# ua_no = ua['No']
# ua_no
ua_Part_Number = ua['Part_Number'] 
ua_Part_Number
```

```python
ua_url = ua['SPEC-Url']
ua_url
```

```python
df = pd.DataFrame(columns=['Part_Number',
#                            'OE',
                           'status'])

df
```

```python
for j in range(2389,2875):
#     print(j)

    url = ua_url[j]
    
    resp = requests.get(url).text
    soup = BeautifulSoup(resp, 'lxml')
    html = etree.HTML(str(soup))
    
    df_temp = pd.DataFrame([{'Part_Number',
                             'OE',
                             'status'}])
    
    list = html.xpath('//tr[@class="awe-row"]/td')
#     print(list[1].text)
#     print(len(list))

    df_temp = pd.DataFrame([{'Part_Number': ua_Part_Number[j],
#                  'OE':list_OE ,
                 'status': 'ok'}])
    
    list_name = [name.text for name in html.xpath('//tr[@class="awe-row"]/td[1]')]
    list_name
    list_value = [name.text for name in html.xpath('//tr[@class="awe-row"]/td[2]')]
    list_value
    
    for z in range(len(list_name)):
        name = list_name[z]
        value = list_value[z]
        df_temp[name.lower()] = value
            

    print(j,ua_Part_Number[j])
    
    print(df_temp) 
    
    df = pd.concat([df, df_temp], ignore_index=True).fillna('')
    
df
```
