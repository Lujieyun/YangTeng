```python
import os
import pandas as pd
import json

list_file = list(os.walk('C:/Users/YangTeng/Downloads/爬虫/aapex'))[0][2]
if '.DS_Store' in list_file:
   list_file.remove('.DS_Store')

len(list_file)

df = pd.DataFrame(columns=['name',
                           'detail',
                           'Url'])

df

for file in list_file:
    print(file)
    with open('C:/Users/YangTeng/Downloads/爬虫/aapex/' + file,encoding='utf-8') as f:
        list_part = f.read()[1:-1].split('},{')
        for i in range(len(list_part)):
            if i == 0:
                part = json.loads(list_part[i] + '}')
            elif i == len(list_part) - 1:
                part = json.loads('{' + list_part[i])
            else:
                part = json.loads('{' + list_part[i] + '}')
                
            df_temp = pd.DataFrame([{'name': part['name'],
                                     'detail':part['detail'],
                                     'Url': part['Url']}])
                                     
            df = pd.concat([df, df_temp], ignore_index=True)
            
df.to_excel('C:/Users/YangTeng/Downloads/爬虫/aapex-menu-1.xlsx', index=False)

df
```
