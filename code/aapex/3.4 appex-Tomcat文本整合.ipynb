```python
import os
list_file = list(os.walk('C:/Users/YangTeng/Downloads/爬虫/aapex/item'))[0][2]
if '.DS_Store' in list_file:
   list_file.remove('.DS_Store')

len(list_file)
```

```python
import pandas as pd
df = pd.DataFrame(columns=['Detail', 'Url'])

df
```

```python
import json
for file in list_file:
    print(file)
    with open('C:/Users/YangTeng/Downloads/爬虫/aapex/item/' + file,encoding='UTF-8') as f:
        part = json.loads(f.read()[1:-1])
        if len(part) != 0:
            df_temp = pd.DataFrame([{'Detail': part['Detail'],
                                     'Url': part['Url']}])
        else:
            df_temp = pd.DataFrame([{'Detail': '',
                                     'Url': ''}])
            
        df = pd.concat([df, df_temp], ignore_index=True)
```

```python
df.to_excel('C:/Users/YangTeng/Downloads/爬虫/aapex/item-detail.xlsx', index=False)

df
```
