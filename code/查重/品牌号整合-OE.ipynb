```python
import pandas as pd
```

```python
source = r'C:\Users\YangTeng\Desktop\工作\品牌号整合\Joanne【模板】.xlsx'# 输入带查重文件路径

out = r'C:\Users\YangTeng\Desktop\工作\品牌号整合\belt_tensioner_NOW().xlsx' # 输入输出文件路径
```

```python
brand_first_data = pd.read_excel(source,sheet_name='品牌1',dtype=str).fillna('')
brand_first_data
```

```python
clean_brand_first_data = brand_first_data[brand_first_data['OE']!='']

clean_brand_first_data
```

```python
new_brand_first_data = clean_brand_first_data[['品牌1','OE']].copy()
new_brand_first_data['OE'] = new_brand_first_data['OE'].str.split(';')
new_brand_first_data = new_brand_first_data.explode('OE')

new_brand_first_data
```

#### ----------------------------------------------------------------
读取品牌2


```python
brand_second_data = pd.read_excel(source,sheet_name='品牌2',dtype=str).fillna('')
brand_second_data
```

```python
clean_brand_second_data = brand_second_data[brand_second_data['OE']!='']

clean_brand_second_data
```

```python
new_brand_second_data = clean_brand_second_data[['品牌2','OE']].copy()
new_brand_second_data['OE'] = new_brand_second_data['OE'].str.split(';')
new_brand_second_data = new_brand_second_data.explode('OE')

new_brand_second_data
```

#### 通过OE连接品牌1与品牌2

```python
merge_data_12 = pd.merge(new_brand_first_data,new_brand_second_data,on='OE',how='left').fillna('')

merge_data_12
```

```python
repeat_data_12 = merge_data_12[merge_data_12['品牌2']!='']

repeat_data_12
```

```python
OE_repeat_data_12 = repeat_data_12.groupby(['品牌1'],as_index=False,sort=False).agg({'OE':';'.join,'品牌2':';'.join})

OE_repeat_data_12
```

```python
OE_repeat_data_12['OE'] = OE_repeat_data_12['OE'].str.split(';').apply(lambda x: sorted(list(set(x)), key=x.index))
OE_repeat_data_12['品牌2'] = OE_repeat_data_12['品牌2'].str.split(';').apply(lambda x: sorted(list((set(x))), key=x.index))
```

```python
OE_repeat_data_12
```

```python
OE_repeat_data_12.rename(columns={'OE':'品牌2重复OE','品牌2':'品牌2'},inplace=True)

OE_repeat_data_12
```

```python
brand_first_data
```

```python
OE_repeat_data_12['品牌2重复OE'] = OE_repeat_data_12['品牌2重复OE'].apply(lambda x: ';'.join(x))
OE_repeat_data_12['品牌2'] = OE_repeat_data_12['品牌2'].apply(lambda x: ';'.join(x))
```

```python
OE_repeat_data_12
```

```python
result_12 = pd.merge(brand_first_data,OE_repeat_data_12,on='品牌1',how='left').fillna('')

result_12
```
