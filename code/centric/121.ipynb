```python
part = html.xpath('//div[@class="product-basic-details__text--part"]/span[1]/text()') 
part
```

```python
Features = html.xpath('//div[@class="product-basic-details__features"]/p[1]/text()') 
Features
```

```python
df_temp = pd.DataFrame([{'Url': url,
                         'part': part,
                         'Features': Features,
                         'status': 'ok'}])

df_temp
```

```python
list_name_1 = [name.text for name in html.xpath('//table[@class="product-basic-details__table"]/descendant::td')]
list_name_1 = [x.strip() for x in list_name_1 if x.strip()]
list_name_1
```

```python
len(list_name_1)
```

```python
q = 0
number_list = []
while q <len(list_name_1):
    
    number_list.append(q)
    
    q=q+2
 
list_name = []
list_value = []
for i in number_list:
    list_name.append(list_name_1[i])
    list_value.append(list_name_1[i+1])
```

```python
for name, value in zip(list_name, list_value):
    df_temp[name.lower()] = value
df_temp
```
