```python
from gevent import monkey
monkey.patch_all()
from gevent.queue import Queue
import gevent

# = = = = = = = = = = = = = = = = = =

import os
import pandas as pd
import requests
import re 
from bs4 import BeautifulSoup
from lxml import etree
import json
from selenium.webdriver import ChromeOptions
import time


# = = = = = = = = = = = = = = = = = =

df_menu = pd.read_excel(r'C:\Users\YangTeng\Desktop\121-img-menu.xlsx',header=0)
list_url = df_menu['Url'].to_list()

print('总数量：' + str(len(list_url)))
print()

```

```python
work = Queue()
for url in list_url:
    work.put_nowait(url)

# = = = = = = = = = = = = = = = = = =

df = pd.DataFrame(columns=['Url',
                           'status'])

# = = = = = = = = = = = = = = = = = =

proxies = {'http': 'http://t19182064816778:9s5s2nv0@f290.kdltps.com:15818',
           'https': 'http://t19182064816778:9s5s2nv0@f290.kdltps.com:15818'}
    
```

```python
def crawler():
    global df
    
    while not work.empty():
        url = work.get_nowait()

        status = 'error'
        for _ in range(31):
            try:
                resp = requests.get(url, proxies=proxies)
                
                part_page = BeautifulSoup(resp.text,'lxml')

                html = etree.HTML(str(part_page))
                part_page
                
                part = html.xpath('//div[@class="product-basic-details__text--part"]/span[1]/text()') 
                
                Features = html.xpath('//div[@class="product-basic-details__features"]/p[1]/text()') 
                
                Application_Summary = html.xpath('//div[@class="product-basic-details__applicable-car"]/text()')
                
                status = 'ok'
                
                df_temp = pd.DataFrame([{'Url': url,
                                         'Features': Features,
                                         'Application_Summary': Application_Summary,
                                         'status': 'ok'}])

                df_temp
                # = = = = = = = = = = = = = = = = = =
                list_name_1 = [name.text for name in html.xpath('//table[@class="product-basic-details__table"]/descendant::td')]
                list_name_1 = [x.strip() for x in list_name_1 if x.strip()]
                list_name_1
                
                len(list_name_1)
                q = 0
                number_list = []
                while q <len(list_name_1):

                    number_list.append(q)

                    q=q+2

                list_name = []
                list_value = []
                for i in number_list:
                    list_name.append(list_name_1[i])
                    list_value.append(list_name_1[i+1])  
                    
                for name, value in zip(list_name, list_value):
                    df_temp[name.lower()] = value
                df_temp
                
                list_name_2 = [name.text for name in html.xpath('//div[@class="product-basic-details__text--section"]/descendant::td')][2:]
                list_name_2 = [x.strip() for x in list_name_2 if x.strip()]
                list_name_2                   
                p = 0
                number_list = []
                while p <len(list_name_2):
                    number_list.append(p)
                    p=p+2

                list_name = []
                list_value = []
                for i in number_list:
                    list_name.append(list_name_2[i])
                    list_value.append(list_name_2[i+1])
                    
                for name, value in zip(list_name, list_value):
                    df_temp[name.lower()] = value
                df_temp            
                

                break
            
            except:
                time.sleep(3)
                continue
                    
        # = = = = = = = = = = = = = = = = = =
        
        if status == 'error':
            df_temp = pd.DataFrame([{'Url': url,
                                     'status': 'error'}])

        df = pd.concat([df, df_temp], ignore_index=True).fillna('')

        print(url +  '  <->  [' + status + '] - 剩余：' + str(work.qsize()))
```

```python
list_task = []
for _ in range(31):
    task = gevent.spawn(crawler)
    list_task.append(task)
gevent.joinall(list_task)

df.to_excel(r'C:\Users\YangTeng\Desktop\121-img-menu-detail(1).xlsx', index=False)
print()
print('搞定')
```
