```python
import os
list_file = list(os.walk('C:/Users/YangTeng/Desktop/文档/爬虫/blower motor resistor'))[0][2]
if '.DS_Store' in list_file:
    list_file.remove('.DS_Store')
if '0.null.txt' in list_file:
    list_file.remove('0.null.txt')
    
len(list_file)
```

```python
import pandas as pd
df = pd.DataFrame()

df
```

```python
from tqdm import tqdm
import json
for file in tqdm(list_file, desc='Progress', ncols=77):
    with open('C:/Users/YangTeng/Desktop/文档/爬虫/blower motor resistor/' + file, 'r', encoding='UTF-8') as f:
        df = pd.concat([df, pd.DataFrame(json.loads(f.read()))], ignore_index=False)
        
df = df[['Part_Number', 'Url']]

df.to_excel('C:/Users/YangTeng/Desktop/文档/爬虫/blower motor resistor.xlsx', index=False)

df
```
