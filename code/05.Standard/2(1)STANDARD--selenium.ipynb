```python
Part_Type = browser.find_element(by=By.XPATH, value='//div[@class="toggle_container_pt"]/ul/li').text
Part_Type
```

```python
Vehicle = browser.find_element(by=By.XPATH, value='//div[@class="toggle_container_bg divBg"]/ul').text
Vehicle
```

```python
df_temp = pd.DataFrame([{'Part_Number':list_part_number[i],
                           'Part_Type':Part_Type,
                           'Vehicle':Vehicle,
                           'status':'ok'}])
df_temp
```

```python
index = browser.find_element(by=By.XPATH, value='//a[@title="Show All Part Specifications"]')

index.click()

list_name = [name.text for name in browser.find_elements(by=By.XPATH, value='//div[@class="spec-group"]/ul/li/span[1]')]

list_value = [value.text for value in browser.find_elements(by=By.XPATH, value='//div[@class="spec-group"]/ul/li/span[2]')]

for name, value in zip(list_name, list_value):
    df_temp[name.lower()] = value
```
