```python
def crawler():
    global df
    
    while not work.empty():
        info = work.get_nowait()
        
        status = 'error'
        for _ in range(31):
            try:
                
                browser = Chrome(service=Service(r'C:\Users\YangTeng\Desktop\venv\chromedriver.exe'), options=option)
                browser.maximize_window()
                
                browser.get(info)

                time.sleep(5)

                list_row = browser.find_elements(by=By.XPATH, value='//table[@class="moreinfotable"]/descendant::tr')[1:]

                status = 'ok'
                
                df_temp = pd.DataFrame([{'Info': info,
                                         'status': 'ok'}])
                for row in list_row:
                    
                    df_temp[row.find_elements(by=By.XPATH, value='./td')[0].text.lower()] = row.find_elements(by=By.XPATH, value='./td')[1].text
                
                browser.quit()
                
                break
                
            except:
                time.sleep(3)
                continue
                
        # = = = = = = = = = = = = = = = = = =

        if status == 'error':
            df_temp = pd.DataFrame([{'Info': info,
                                     'status': 'error'}])
        df = pd.concat([df, df_temp], ignore_index=True).fillna('')
        
        print(info +  '  <->  [' + status + '] - 剩余数量：' + str(work.qsize()))
        
```
