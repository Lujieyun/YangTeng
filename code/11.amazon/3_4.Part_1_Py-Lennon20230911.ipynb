```python
import os
list_file = list(os.walk(r'C:\Users\YangTeng\Downloads\爬虫\amazon\销量-1'))[0][2]
if '.DS_Store' in list_file:
    list_file.remove('.DS_Store')
if '0.null.txt' in list_file:
    list_file.remove('0.null.txt')
    
len(list_file)
```

```python
import pandas as pd
df = pd.DataFrame()

df
```

```python
from tqdm import tqdm
import json
for file in tqdm(list_file, desc='进度', ncols=77):
    with open('C:/Users/YangTeng/Downloads/爬虫/amazon/销量-1/' + file, 'r', encoding='UTF-8') as f:
        df = pd.concat([df, pd.DataFrame(json.loads(f.read()))], ignore_index=False)

df.to_excel(r'C:\Users\YangTeng\Downloads\爬虫\amazon\销量\销量-1.xlsx', index=False)

df
```
