```python
import os
list_file = list(os.walk(r'C:\Users\YangTeng\Downloads\爬虫\爬\MAF\CARDONE'))[0][2]
if '.DS_Store' in list_file:
   list_file.remove('.DS_Store')

len(list_file)
```

```python
import pandas as pd
df = pd.DataFrame(columns=['Title', 'Vehicle', 'Url'])

df
```

```python
import json
for file in list_file:
    print(file)
    with open('C:/Users/YangTeng/Downloads/爬虫/爬/MAF/CARDONE/' + file) as f:
        list_part = f.read()[1:-1].split('},{')
        for i in range(len(list_part)):
            if i == 0:
                part = json.loads(list_part[i] + '}')
            elif i == len(list_part) - 1:
                part = json.loads('{' + list_part[i])
            else:
                part = json.loads('{' + list_part[i] + '}')
                
            df_temp = pd.DataFrame([{'Title': part['Title'],
                                     'Vehicle': part['Vehicle'],
                                     'Url': part['Url']}])
            df = pd.concat([df, df_temp], ignore_index=True)
```
